## Feature Engineering

You may want to convert categorical variables to numerical. For example, education takes on the value Graduate and Not Graduate. But we want it to be 0 or 1 for machine learning algorithms to use.

```python
loan_data['education'] = loan_data['education'].map({"Graduate": 1, 'Not Graduate': 0}).fillna(0).astype(int)
loan_data['self_employed'] = loan_data['self_employed'].map({"Yes": 1, 'No': 0}).fillna(0).astype(int)
loan_data['loan_status'] = loan_data['loan_status'].map({"Approved": 1, 'Rejected': 0}).fillna(0).astype(int)

# Hint: Other categorical variables are self_employed and loan_status
# TODO
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

## Model Training and Evaluation

```python
X = loan_data.select_dtypes(include=['float64', 'int64'])

# Standardize the features
scaler = StandardScaler()
scaled_data = scaler.fit_transform(X)
```

```python
pca = PCA(n_components=3)  # Keep 2 components for a 2D map
pca_features = pca.fit_transform(scaled_data)
```

```python
import seaborn as sns
```

```python
import plotly.express as px

var_1, var_2, var_3 = pca.explained_variance_ratio_ * 100

pca_plot_data = pd.DataFrame({
    'PC1': pca_features[:, 0],
    'PC2': pca_features[:, 1],
    'PC3': pca_features[:, 2],
    'Loan Status': loan_data['loan_status'].map({0: 'Rejected', 1: 'Approved'})
})

fig = px.scatter_3d(
    pca_plot_data,
    x='PC1',
    y='PC2',
    z='PC3',
    color='Loan Status',
    color_discrete_map={'Rejected': '#d95f02', 'Approved': '#1b9e77'},
    hover_data={'PC1': ':.2f', 'PC2': ':.2f', 'PC3': ':.2f'},
    title='Interactive 3D PCA Map of Loan Applications',
    labels={
        'PC1': f'PC1 ({var_1:.1f}% variance)',
        'PC2': f'PC2 ({var_2:.1f}% variance)',
        'PC3': f'PC3 ({var_3:.1f}% variance)'
    }
)

fig.update_traces(marker={'size': 4, 'opacity': 0.75})
fig.update_layout(
    height=700,
    legend_title_text='Loan Status',
    margin={'l': 0, 'r': 0, 'b': 0, 't': 50}
)
fig.show()
```
